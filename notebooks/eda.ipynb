```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("../dataset/breast_cancer.csv")

print("Dataset Shape:", df.shape)

display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
# Missing values
print("Missing Values:")
print(df.isnull().sum().sum())

# Duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Unique target values
print("\nTarget Classes:")
print(df["diagnosis"].unique())

# Target distribution
print("\nTarget Distribution:")
print(df["diagnosis"].value_counts())

# Target percentage
print("\nTarget Percentage:")
print(
    df["diagnosis"].value_counts(normalize=True) * 100
)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="diagnosis"
)

plt.title("Distribution of Breast Cancer Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Records")

plt.show()
```

```python
features = df.drop(columns=["diagnosis"])

features.iloc[:, :10].hist(
    figsize=(16, 12),
    bins=25
)

plt.suptitle("Distribution of Mean Features")
plt.tight_layout()
plt.show()
```

```python
features.iloc[:, 10:20].hist(
    figsize=(16, 12),
    bins=25
)

plt.suptitle("Distribution of Standard Error Features")
plt.tight_layout()
plt.show()
```

```python
features.iloc[:, 20:30].hist(
    figsize=(16, 12),
    bins=25
)

plt.suptitle("Distribution of Worst Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(22, 18))

correlation = features.corr()

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Correlation Heatmap of All 30 Features")
plt.tight_layout()
plt.show()
```

```python
selected_features = [
    "radius_mean",
    "texture_mean",
    "perimeter_mean",
    "area_mean",
    "concavity_mean",
    "concave_points_mean"
]

fig, axes = plt.subplots(2, 3, figsize=(16, 10))

axes = axes.flatten()

for i, column in enumerate(selected_features):

    sns.boxplot(
        data=df,
        x="diagnosis",
        y=column,
        ax=axes[i]
    )

    axes[i].set_title(f"{column} vs Diagnosis")

plt.tight_layout()
plt.show()
```
